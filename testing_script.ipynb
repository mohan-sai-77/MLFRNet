```python
import copy
import math
import os
import shutil
import time
from dataclasses import dataclass, field
from pathlib import Path
from typing import Callable, Dict, List, Optional, Sequence, Tuple

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import psutil
import seaborn as sns
import torch
import torch.nn as nn
from sklearn.metrics import (
    accuracy_score,
    auc,
    average_precision_score,
    cohen_kappa_score,
    confusion_matrix,
    f1_score,
    precision_recall_curve,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.preprocessing import label_binarize
from thop import profile
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from tqdm.auto import tqdm

# ----------------------------------------------------------------------------
# Configuration
# ----------------------------------------------------------------------------
checkpointlist_224 = [
    "/gdrive/MyDrive/JR_1/checkpoint/ConvNext_Complete_classifier_02.pt",
    "/gdrive/MyDrive/complete_35_EPOCHS_model.pt",
]
checkpointlist_384 = ["/gdrive/MyDrive/JR_1/ConvNext_Complete_classifier_res384.pt"]
checkpointlist_512 = ["/gdrive/MyDrive/JR_1/ConvNext_Complete_classifier_res512.pt"]

CHECKPOINT_LISTS: Dict[int, List[str]] = {
    224: checkpointlist_224,
    384: checkpointlist_384,
    512: checkpointlist_512,
}


@dataclass(frozen=True)
class Config:
    test_path_drive: str = (
        "/gdrive/MyDrive/JR_1/Her_bracs_MIHIC_Ostro/data/HER2-IHC-40x/"
        "her2-ihc-40x-patch/Patch-based-dataset/test_data_patch"
    )
    test_path_local: str = "/content/data_cache/test_data_patch"
    use_local_cache: bool = True
    batch_size: int = 32
    num_workers: int = 2
    num_classes: int = 4
    mean: Tuple[float, ...] = (0.485, 0.456, 0.406)
    std: Tuple[float, ...] = (0.229, 0.224, 0.225)
    n_warmup: int = 20
    n_iters: int = 100
    n_ece_bins: int = 15
    resolutions: Optional[Sequence[int]] = None  # None = all keys of CHECKPOINT_LISTS
    show_plots: bool = True
    results_csv: str = "/content/checkpoint_summary.csv"


CFG = Config()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


# ----------------------------------------------------------------------------
# Model definition (required to unpickle the saved .pt / .pth files)
# ----------------------------------------------------------------------------
class EfficientChannelAttention(nn.Module):
    def __init__(self, channels, gamma=2, b=1):
        super().__init__()
        kernel_size = int(abs((math.log2(channels) + b) / gamma))
        kernel_size = kernel_size if kernel_size % 2 else kernel_size + 1
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.conv = nn.Conv1d(
            1, 1, kernel_size=kernel_size, padding=(kernel_size - 1) // 2, bias=False
        )
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        y = self.avg_pool(x)
        y = y.squeeze(-1).transpose(-1, -2)
        y = self.conv(y).transpose(-1, -2).unsqueeze(-1)
        return x * self.sigmoid(y).expand_as(x)


class MultiScaleBlock(nn.Module):
    def __init__(self, in_channels):
        super().__init__()
        self.conv3 = nn.Conv2d(in_channels, in_channels, 3, padding=1, groups=in_channels, bias=False)
        self.conv5 = nn.Conv2d(in_channels, in_channels, 5, padding=2, groups=in_channels, bias=False)
        self.conv7 = nn.Conv2d(in_channels, in_channels, 7, padding=3, groups=in_channels, bias=False)
        self.bn = nn.BatchNorm2d(in_channels * 3)
        self.fusion = nn.Sequential(
            nn.Conv2d(in_channels * 3, in_channels, 1, bias=False),
            nn.BatchNorm2d(in_channels),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        x_cat = torch.cat([self.conv3(x), self.conv5(x), self.conv7(x)], dim=1)
        return self.fusion(self.bn(x_cat)) + x


class RefinementBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.conv1 = nn.Conv2d(channels, channels, 3, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(channels)
        self.conv2 = nn.Conv2d(channels, channels, 3, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(channels)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x):
        identity = x
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.bn2(self.conv2(x))
        return self.relu(x + identity)


class ConvNeXt_Enhanced(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.backbone = models.convnext_tiny(weights=None)
        self.features = self.backbone.features
        self.attention = EfficientChannelAttention(768)
        self.multiscale = MultiScaleBlock(768)
        self.refine = RefinementBlock(768)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.BatchNorm1d(768),
            nn.Dropout(0.4),
            nn.Linear(768, 512),
            nn.GELU(),
            nn.BatchNorm1d(512),
            nn.Dropout(0.3),
            nn.Linear(512, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        x = self.attention(x)
        x = self.multiscale(x)
        x = self.refine(x)
        x = self.pool(x)
        return self.classifier(x)


# ----------------------------------------------------------------------------
# Data
# ----------------------------------------------------------------------------
class DatasetProvider:
    """Stages the test set on local disk once and builds per-resolution loaders."""

    def __init__(self, cfg: Config):
        self.cfg = cfg
        self.root = self._stage()

    def _stage(self) -> str:
        cfg = self.cfg
        if not cfg.use_local_cache:
            return cfg.test_path_drive
        dst = Path(cfg.test_path_local)
        if dst.exists() and any(dst.iterdir()):
            return str(dst)
        dst.parent.mkdir(parents=True, exist_ok=True)
        shutil.copytree(cfg.test_path_drive, dst)
        return str(dst)

    def loader(self, img_size: int) -> Tuple[DataLoader, List[str]]:
        cfg = self.cfg
        tfm = transforms.Compose([
            transforms.Resize((img_size, img_size)),
            transforms.ToTensor(),
            transforms.Normalize(cfg.mean, cfg.std),
        ])
        dataset = datasets.ImageFolder(self.root, transform=tfm)
        loader = DataLoader(
            dataset,
            batch_size=cfg.batch_size,
            shuffle=False,
            num_workers=cfg.num_workers,
            pin_memory=DEVICE.type == "cuda",
        )
        return loader, dataset.classes


# ----------------------------------------------------------------------------
# Inference
# ----------------------------------------------------------------------------
ForwardFn = Callable[[nn.Module, torch.Tensor], torch.Tensor]


def plain_forward(model: nn.Module, images: torch.Tensor) -> torch.Tensor:
    return torch.softmax(model(images), dim=1)


def tta_forward(model: nn.Module, images: torch.Tensor) -> torch.Tensor:
    """Average softmax over identity, horizontal, vertical and both flips."""
    views = (
        images,
        torch.flip(images, dims=[3]),
        torch.flip(images, dims=[2]),
        torch.flip(images, dims=[2, 3]),
    )
    return sum(torch.softmax(model(v), dim=1) for v in views) / len(views)


# Normal test = single pass on the unaugmented image; tta = flip-averaged pass.
EVAL_MODES: Dict[str, ForwardFn] = {
    "normal": plain_forward,
    "tta": tta_forward,
}


@torch.no_grad()
def predict(model: nn.Module, loader: DataLoader, forward_fn: ForwardFn, desc: str):
    probs, labels = [], []
    for images, targets in tqdm(loader, desc=desc, leave=False):
        probs.append(forward_fn(model, images.to(DEVICE, non_blocking=True)).cpu().numpy())
        labels.append(targets.numpy())
    probs = np.concatenate(probs)
    return np.concatenate(labels), probs.argmax(axis=1), probs


# ----------------------------------------------------------------------------
# Model complexity and efficiency
# ----------------------------------------------------------------------------
class ModelProfiler:
    """Parameter counts, MACs/FLOPs, latency, throughput and peak memory."""

    def __init__(self, cfg: Config):
        self.cfg = cfg

    @staticmethod
    def complexity(model: nn.Module, path: str, img_size: int) -> dict:
        total = sum(p.numel() for p in model.parameters())
        trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
        result = {
            "total_params": total,
            "trainable_params": trainable,
            "non_trainable_params": total - trainable,
            "file_size_mb": os.path.getsize(path) / 1024 ** 2,
        }
        try:
            clone = copy.deepcopy(model).eval()
            dummy = torch.randn(1, 3, img_size, img_size, device=DEVICE)
            macs, _ = profile(clone, inputs=(dummy,), verbose=False)
            result.update(macs_g=macs / 1e9, flops_g=2 * macs / 1e9)
            del clone
        except Exception as exc:
            result.update(macs_g=float("nan"), flops_g=float("nan"))
            print(f"FLOPs measurement skipped: {exc}")
        return result

    def latency(self, model, img_size: int, forward_fn: ForwardFn) -> dict:
        cfg, is_cuda = self.cfg, DEVICE.type == "cuda"
        results = {}
        model.eval()
        for bs in (1, cfg.batch_size):
            dummy = torch.randn(bs, 3, img_size, img_size, device=DEVICE)
            timings = []
            with torch.no_grad():
                for _ in range(cfg.n_warmup):
                    forward_fn(model, dummy)
                if is_cuda:
                    torch.cuda.synchronize()
                for _ in range(cfg.n_iters):
                    if is_cuda:
                        start = torch.cuda.Event(enable_timing=True)
                        end = torch.cuda.Event(enable_timing=True)
                        start.record()
                        forward_fn(model, dummy)
                        end.record()
                        torch.cuda.synchronize()
                        timings.append(start.elapsed_time(end))
                    else:
                        t0 = time.perf_counter()
                        forward_fn(model, dummy)
                        timings.append((time.perf_counter() - t0) * 1000)
            mean_ms = float(np.mean(timings))
            results[f"batch_{bs}"] = {
                "mean_ms": mean_ms,
                "std_ms": float(np.std(timings)),
                "imgs_per_sec": bs / (mean_ms / 1000),
            }
        return results

    def memory(self, model, img_size: int, forward_fn: ForwardFn) -> dict:
        bs = self.cfg.batch_size
        model.eval()
        with torch.no_grad():
            if DEVICE.type == "cuda":
                torch.cuda.empty_cache()
                torch.cuda.reset_peak_memory_stats(DEVICE)
                forward_fn(model, torch.randn(bs, 3, img_size, img_size, device=DEVICE))
                torch.cuda.synchronize()
                return {
                    "allocated_mb": torch.cuda.max_memory_allocated(DEVICE) / 1e6,
                    "reserved_mb": torch.cuda.max_memory_reserved(DEVICE) / 1e6,
                }
            process = psutil.Process(os.getpid())
            before = process.memory_info().rss / 1e6
            forward_fn(model, torch.randn(bs, 3, img_size, img_size))
            return {"rss_delta_mb": process.memory_info().rss / 1e6 - before}

    def efficiency(self, model, img_size: int, forward_fn: ForwardFn) -> dict:
        return {
            "latency": self.latency(model, img_size, forward_fn),
            "memory": self.memory(model, img_size, forward_fn),
        }


# ----------------------------------------------------------------------------
# Metrics
# ----------------------------------------------------------------------------
class MetricsCalculator:
    """Discrimination, per-class and calibration metrics from probabilities."""

    def __init__(self, n_ece_bins: int):
        self.n_ece_bins = n_ece_bins

    def ece(self, y_true, y_probs) -> Tuple[float, List[dict]]:
        """Expected calibration error over equal-width top-1 confidence bins."""
        conf = y_probs.max(axis=1)
        correct = (y_probs.argmax(axis=1) == y_true).astype(float)
        edges = np.linspace(0, 1, self.n_ece_bins + 1)
        total, bins = 0.0, []
        for i in range(self.n_ece_bins):
            lo, hi = edges[i], edges[i + 1]
            mask = (conf >= lo) & (conf <= hi) if i == 0 else (conf > lo) & (conf <= hi)
            n = int(mask.sum())
            if n:
                acc_b, conf_b = correct[mask].mean(), conf[mask].mean()
                total += n / len(y_true) * abs(acc_b - conf_b)
                bins.append({"confidence": float(conf_b), "accuracy": float(acc_b), "n": n})
        return float(total), bins

    @staticmethod
    def _per_class(cm, y_bin, y_probs, class_names) -> Tuple[dict, List[float]]:
        per_class, specificities = {}, []
        for i, name in enumerate(class_names):
            tp = int(cm[i, i])
            fn = int(cm[i].sum() - tp)
            fp = int(cm[:, i].sum() - tp)
            tn = int(cm.sum() - tp - fn - fp)
            sens = tp / (tp + fn) if tp + fn else 0.0
            spec = tn / (tn + fp) if tn + fp else 0.0
            prec = tp / (tp + fp) if tp + fp else 0.0
            f1 = 2 * prec * sens / (prec + sens) if prec + sens else 0.0
            try:
                auroc = roc_auc_score(y_bin[:, i], y_probs[:, i])
            except ValueError:
                auroc = float("nan")
            specificities.append(spec)
            per_class[name] = {
                "sensitivity": sens, "specificity": spec, "precision": prec, "f1": f1,
                "auroc": auroc,
                "pr_auc": average_precision_score(y_bin[:, i], y_probs[:, i]),
            }
        return per_class, specificities

    def compute(self, y_true, y_pred, y_probs, class_names) -> dict:
        n_classes = y_probs.shape[1]
        labels = list(range(n_classes))
        y_bin = label_binarize(y_true, classes=labels)
        cm = confusion_matrix(y_true, y_pred, labels=labels)
        per_class, specificities = self._per_class(cm, y_bin, y_probs, class_names)

        fpr, tpr, _ = roc_curve(y_bin.ravel(), y_probs.ravel())
        ece, ece_bins = self.ece(y_true, y_probs)

        overall = {"accuracy": accuracy_score(y_true, y_pred)}
        for avg in ("macro", "micro", "weighted"):
            overall[f"precision_{avg}"] = precision_score(y_true, y_pred, average=avg, zero_division=0)
            overall[f"recall_{avg}"] = recall_score(y_true, y_pred, average=avg, zero_division=0)
            overall[f"f1_{avg}"] = f1_score(y_true, y_pred, average=avg, zero_division=0)
            overall[f"pr_auc_{avg}"] = average_precision_score(y_bin, y_probs, average=avg)
        for avg in ("macro", "weighted"):
            overall[f"auroc_{avg}"] = roc_auc_score(y_bin, y_probs, average=avg, multi_class="ovr")
        overall.update(
            auroc_micro=float(auc(fpr, tpr)),
            specificity_macro=float(np.mean(specificities)),
            cohen_kappa=cohen_kappa_score(y_true, y_pred),
            ece=ece,
            brier=float(np.mean(np.sum((y_probs - y_bin) ** 2, axis=1))),
        )
        return {
            "overall": overall,
            "per_class": per_class,
            "confusion_matrix": cm,
            "ece_bins": ece_bins,
            "y_bin": y_bin,
        }


# ----------------------------------------------------------------------------
# Reporting and visualisation
# ----------------------------------------------------------------------------
class Reporter:
    @staticmethod
    def print_complexity(name, complexity):
        print(f"\n{name}")
        print(
            f"params {complexity['total_params']:,} "
            f"(trainable {complexity['trainable_params']:,}) | "
            f"file {complexity['file_size_mb']:.1f} MB | "
            f"MACs {complexity['macs_g']:.2f} G | FLOPs {complexity['flops_g']:.2f} G"
        )

    @staticmethod
    def print_mode(mode, eff, quality):
        print(f"\n[{mode}]")
        for key, v in eff["latency"].items():
            print(f"{key}: {v['mean_ms']:.2f} +/- {v['std_ms']:.2f} ms, {v['imgs_per_sec']:.1f} img/s")
        print("memory: " + ", ".join(f"{k} {v:.1f} MB" for k, v in eff["memory"].items()))

        o = quality["overall"]
        print(f"accuracy {o['accuracy']:.4f} | kappa {o['cohen_kappa']:.4f} | "
              f"ECE {o['ece']:.4f} | Brier {o['brier']:.4f}")
        rows = {
            metric: [o[f"{metric}_{avg}"] for avg in ("macro", "micro", "weighted")]
            for metric in ("precision", "recall", "f1", "auroc", "pr_auc")
        }
        print(pd.DataFrame(rows, index=["macro", "micro", "weighted"]).T.round(4).to_string())
        print(pd.DataFrame(quality["per_class"]).T.round(4).to_string())

    @staticmethod
    def summary_row(name, res, mode, complexity, eff, quality) -> dict:
        o = quality["overall"]
        return {
            "checkpoint": name,
            "resolution": res,
            "mode": mode,
            "accuracy": o["accuracy"],
            "f1_macro": o["f1_macro"],
            "auroc_macro": o["auroc_macro"],
            "kappa": o["cohen_kappa"],
            "ece": o["ece"],
            "params_m": complexity["total_params"] / 1e6,
            "flops_g": complexity["flops_g"],
            "bs1_ms": eff["latency"]["batch_1"]["mean_ms"],
        }


class Plotter:
    @staticmethod
    def confusion(cm, class_names, title):
        plt.figure(figsize=(6, 5))
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                    xticklabels=class_names, yticklabels=class_names)
        plt.title(f"Confusion matrix: {title}")
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.tight_layout()
        plt.show()

    @staticmethod
    def sens_spec(per_class, class_names):
        x, w = np.arange(len(class_names)), 0.35
        plt.figure(figsize=(7, 5))
        plt.bar(x - w / 2, [per_class[c]["sensitivity"] for c in class_names], w, label="Sensitivity")
        plt.bar(x + w / 2, [per_class[c]["specificity"] for c in class_names], w, label="Specificity")
        plt.xticks(x, class_names)
        plt.ylim(0, 1.05)
        plt.ylabel("Score")
        plt.title("Per-class sensitivity and specificity")
        plt.legend()
        plt.tight_layout()
        plt.show()

    @staticmethod
    def roc(y_bin, y_probs, class_names):
        n = len(class_names)
        fpr, tpr = {}, {}
        plt.figure(figsize=(7, 6))
        for i in range(n):
            fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], y_probs[:, i])
            plt.plot(fpr[i], tpr[i], label=f"{class_names[i]} (AUC={auc(fpr[i], tpr[i]):.3f})")
        f_micro, t_micro, _ = roc_curve(y_bin.ravel(), y_probs.ravel())
        grid = np.unique(np.concatenate([fpr[i] for i in range(n)]))
        t_macro = np.mean([np.interp(grid, fpr[i], tpr[i]) for i in range(n)], axis=0)
        plt.plot(f_micro, t_micro, ":", label=f"micro (AUC={auc(f_micro, t_micro):.3f})")
        plt.plot(grid, t_macro, "--", label=f"macro (AUC={auc(grid, t_macro):.3f})")
        plt.plot([0, 1], [0, 1], "k--", linewidth=0.8)
        plt.xlabel("False positive rate")
        plt.ylabel("True positive rate")
        plt.title("ROC curves")
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

    @staticmethod
    def pr(y_bin, y_probs, class_names, per_class, pr_auc_micro):
        plt.figure(figsize=(7, 6))
        for i, name in enumerate(class_names):
            p, r, _ = precision_recall_curve(y_bin[:, i], y_probs[:, i])
            plt.plot(r, p, label=f"{name} (AP={per_class[name]['pr_auc']:.3f})")
        p, r, _ = precision_recall_curve(y_bin.ravel(), y_probs.ravel())
        plt.plot(r, p, ":", label=f"micro (AP={pr_auc_micro:.3f})")
        plt.xlabel("Recall")
        plt.ylabel("Precision")
        plt.title("Precision-recall curves")
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

    @staticmethod
    def reliability(ece_bins, ece_value):
        plt.figure(figsize=(6, 6))
        plt.bar([b["confidence"] for b in ece_bins], [b["accuracy"] for b in ece_bins],
                width=0.05, alpha=0.75, edgecolor="black", label="Model")
        plt.plot([0, 1], [0, 1], "k--", label="Perfect calibration")
        plt.xlabel("Confidence")
        plt.ylabel("Accuracy")
        plt.title(f"Reliability diagram (ECE = {ece_value:.4f})")
        plt.xlim(0, 1)
        plt.ylim(0, 1)
        plt.legend()
        plt.tight_layout()
        plt.show()

    @classmethod
    def all(cls, title, y_probs, class_names, quality):
        cls.confusion(quality["confusion_matrix"], class_names, title)
        cls.sens_spec(quality["per_class"], class_names)
        cls.roc(quality["y_bin"], y_probs, class_names)
        cls.pr(quality["y_bin"], y_probs, class_names, quality["per_class"],
               quality["overall"]["pr_auc_micro"])
        cls.reliability(quality["ece_bins"], quality["overall"]["ece"])


# ----------------------------------------------------------------------------
# Orchestration
# ----------------------------------------------------------------------------
class Evaluator:
    def __init__(self, cfg: Config, checkpoint_lists: Dict[int, List[str]]):
        self.cfg = cfg
        self.checkpoint_lists = checkpoint_lists
        self.data = DatasetProvider(cfg)
        self.profiler = ModelProfiler(cfg)
        self.metrics = MetricsCalculator(cfg.n_ece_bins)

    STATE_DICT_KEYS = ("model_state_dict", "state_dict", "model", "net", "network")

    def load_model(self, path: str) -> nn.Module:
        """Load either a fully pickled model or a state-dict style checkpoint."""
        obj = torch.load(path, map_location=DEVICE, weights_only=False)

        if isinstance(obj, dict):
            state = obj
            for key in self.STATE_DICT_KEYS:
                if key in obj:
                    state = obj[key]
                    break
            if isinstance(state, nn.Module):
                obj = state
            else:
                state = {k.removeprefix("module."): v for k, v in state.items()}
                obj = ConvNeXt_Enhanced(self.cfg.num_classes)
                obj.load_state_dict(state, strict=True)

        return obj.to(DEVICE).eval()

    def evaluate_checkpoint(self, path: str, res: int, loader, class_names) -> List[dict]:
        model = self.load_model(path)
        name = Path(path).name

        complexity = self.profiler.complexity(model, path, res)
        Reporter.print_complexity(f"{name} ({res}px)", complexity)

        rows = []
        for mode, forward_fn in EVAL_MODES.items():
            eff = self.profiler.efficiency(model, res, forward_fn)
            y_true, y_pred, y_probs = predict(model, loader, forward_fn, f"{name} {mode}")
            quality = self.metrics.compute(y_true, y_pred, y_probs, class_names)

            Reporter.print_mode(mode, eff, quality)
            if self.cfg.show_plots:
                Plotter.all(f"{name} ({res}px, {mode})", y_probs, class_names, quality)
            rows.append(Reporter.summary_row(name, res, mode, complexity, eff, quality))

        del model
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()
        return rows

    def run(self) -> pd.DataFrame:
        resolutions = self.cfg.resolutions or list(self.checkpoint_lists)
        rows = []
        for res in resolutions:
            loader, class_names = self.data.loader(res)
            for path in tqdm(self.checkpoint_lists[res], desc=f"{res}px checkpoints"):
                if not os.path.isfile(path):
                    print(f"missing: {path}")
                    continue
                rows.extend(self.evaluate_checkpoint(path, res, loader, class_names))

        summary = pd.DataFrame(rows)
        if not summary.empty:
            summary.to_csv(self.cfg.results_csv, index=False)
            print("\nSummary")
            print(summary.round(4).to_string(index=False))
        return summary


# ----------------------------------------------------------------------------
# Entry point
# ----------------------------------------------------------------------------
print(f"device: {DEVICE}" + (f" ({torch.cuda.get_device_name(0)})" if DEVICE.type == "cuda" else ""))
summary_df = Evaluator(CFG, CHECKPOINT_LISTS).run()
```
